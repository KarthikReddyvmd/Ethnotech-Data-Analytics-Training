```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statistics
import seaborn as sns
```

```python
df = pd.read_excel('Customer_Churn.xlsx')
```

```python
df
```

```python
df.loc[df['Churn'] == 'No', 'Churn'] = 0
df.loc[df['Churn'] == 'Yes', 'Churn'] = 1
```

```python
df
```

```python
df['Churn'].value_counts()//len(df)*100
```

```python
x=df.drop(['customerID','Churn'],axis = 1)
```

```python
x
```

```python
y = df.Churn.values
```

```python
y
```

```python
x.columns
```

```python
# CONVERT CATEGORICAL FEATURES TO NUMERICAL FEATURES
```

```python
x = pd.get_dummies(x, columns = ['gender', 'Partner', 'Dependents', 'PhoneService', 'MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies', 'Contract', 'PaperlessBilling', 'PaymentMethod'] , drop_first = True)
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2, random_state = 0)
```

```python
len(x_train)
```

```python
len(x_test)
```

```python
# Feature Scaling
```

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

sc = StandardScaler()

# Convert 'TotalCharges' to numeric, handling errors by coercing non-numeric values to NaN
x_train['TotalCharges'] = pd.to_numeric(x_train['TotalCharges'], errors='coerce')
x_test['TotalCharges'] = pd.to_numeric(x_test['TotalCharges'], errors='coerce')

# Fill any NaN values introduced by coercion with 0
x_train['TotalCharges'] = x_train['TotalCharges'].fillna(0)
x_test['TotalCharges'] = x_test['TotalCharges'].fillna(0)

x_train_sc = sc.fit_transform(x_train)
x_test_sc = sc.transform(x_test)

```

```python
x_train_sc
```

```python
x_test_sc
```

KNN - K Nearest Neighbours

```python
from sklearn.neighbors import KNeighborsClassifier
import numpy as np

# Convert y_train and y_test to integer type
y_train = y_train.astype(int)
y_test = y_test.astype(int)

model = KNeighborsClassifier(n_neighbors = 5)
model.fit(x_train_sc, y_train)
```

```python
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
y_pred = model.predict(x_test_sc)
cm = confusion_matrix(y_test, y_pred)
print(cm)
```

```python
print(classification_report(y_test, y_pred))
```

```python
print(accuracy_score(y_test, y_pred))
```

DECISION TREE

```python
from sklearn.tree import DecisionTreeClassifier
model_dt = DecisionTreeClassifier()
model_dt.fit(x_train_sc, y_train)
```

```python
y_pred = model_dt.predict(x_test_sc)
cm = confusion_matrix(y_test, y_pred)
print(cm)
```
