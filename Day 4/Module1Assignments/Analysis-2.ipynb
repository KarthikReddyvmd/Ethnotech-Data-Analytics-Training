```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
import statistics

Product_purchased=['Laptop','Mobile','Laptop','Earphones','Laptop','Mobile','Backpack','Laptop','Book','Mobile','Earphones','Mobile','Earphones','Laptop']
Cart_abandoment=['Yes','No','Yes','Yes','No','No','Yes','Yes','No','Yes','No','Yes','Yes','No']
Customer_Ratings=[4.5,4,4.8,5,4.1,3.9,4,3.8,4.7,4.2,4,3.9,4.1,5]
Payment_Methods=['UPI','NetBanking','CashOnDelivery','UPI','UPI','CashOnDelivery','UPI','NetBanking','CashOnDelivery','UPI','UPI','CashOnDelivery','UPI','CashOnDelivery']

df=pd.DataFrame({'Product_purchased':Product_purchased,'Cart_abandoment':Cart_abandoment,'Customer_Ratings':Customer_Ratings,'Payment_Methods':Payment_Methods})
df
```

```python
print('---------------Most Sold Products----------------')
print(df['Product_purchased'].value_counts())
print('\n')
print('Most Sold Product:  ',df['Product_purchased'].value_counts().idxmax())
print('\n')

plt.figure(figsize=(10,5))
plt.pie(df['Product_purchased'].value_counts(),labels=df['Product_purchased'].value_counts().index,autopct='%1.1f%%')
plt.title('Most Sold Products')
plt.show()
```

```python
print('---------------High Revenue Customers----------------')
print('/n',df['Product_purchased'].value_counts())
print('\n')
print('High Revenue Product:  ',df['Product_purchased'].value_counts().idxmax())
print('\n')
print('High Revenue Customer:  ',df['Product_purchased'].value_counts().max())

plt.figure(figsize=(10,5))
sb.countplot(data=df,x='Product_purchased',color='skyblue')
plt.title('High Revenue Customers')
plt.show()
```

```python
print('--------------Most used Payment Methods----------------')
print(df['Payment_Methods'].value_counts())
print('\n')
print('Most used Payment Method:  ',df['Payment_Methods'].value_counts().idxmax())
print('\n')
```
