```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
import statistics

Order_time=[6,8,12,1,6,2,9,10,2,5]
Delivery_time=[30,20,39,45,60,40,35,55,65,40]
Delivery_partner=[1,3,2,1,2,3,1,1,2,3]
Customer_ratings=[5,4,5,2,2,3,5,4,5,3]

df=pd.DataFrame({'Order_time':Order_time,'Delivery_time':Delivery_time,'Delivery_partner':Delivery_partner,'Customer_ratings':Customer_ratings})
df
```

```python
print('--------------Average Delivary Time-------------------')
print('\n',df.groupby('Delivery_partner')['Delivery_time'].mean())
```

```python
print('---------------Peak Order Hours-------------------')

order_counts = df['Order_time'].value_counts().sort_index()

print('\nMaximum Order Time:', df['Order_time'].max())

plt.figure(figsize=(10, 6))
plt.stem(order_counts.index, order_counts.values)
plt.title('Frequency of Orders by Hour')
plt.xlabel('Order Hour')
plt.ylabel('Number of Orders')
plt.xticks(order_counts.index)
plt.grid(axis='y', linestyle='--')
plt.show()
```

```python
print('---------------Most Delayed Areas--------------')
print('\n',df.groupby('Delivery_partner')['Delivery_time'].max())

plt.figure(figsize=(10, 6))
plt.bar(df['Delivery_partner'], df['Delivery_time'])
plt.title('Delivery Time by Delivery Partner')
plt.xlabel('Delivery Partner')
plt.ylabel('Delivery Time')
```

```python
print('---------------Top rated Delivery partners-------------------')
print('\n',df.groupby('Delivery_partner')['Customer_ratings'].mean())

plt.figure(figsize=(10,5))
plt.bar(df['Delivery_partner'],df['Customer_ratings'])
plt.title('Customer Ratings by Delivery Partner')
plt.xlabel('Delivery Partner')
plt.ylabel('Customer Ratings')
plt.show()
```
