```python
import numpy as np
import pandas as pd
import statistics
import matplotlib.pyplot as plt
import seaborn as sb
Customers=["Amar","Bob","Jhon","Steev","Karan","Rahul","Kiran","Aman"]
Internet_Speed=[5,9,6,14,3,15,8,2]
Payment_Failure=[1,0,1,1,0,1,1,0]
Delivery_Charges=[30,50,90,20,50,100,40,70]
Purchase_Completed=[1,1,0,1,1,0,1,0]

print("--------------Custoer Details-----------------")
df=pd.DataFrame({"Customers":Customers,"Internet_Speed":Internet_Speed,"Payment_Failure":Payment_Failure,"Delivery_Charges":Delivery_Charges,"Purchase_Completed":Purchase_Completed})
print(df)

print("----------------Total Abandmont vs Completed Abandont-----------------")
sb.countplot(x='Purchase_Completed',data=df)
plt.show()
counts=df['Purchase_Completed'].value_counts()
plt.bar(counts.index,counts.values,color=['red','green'])
plt.xticks([0,1],['Abandoned','Completed'])
plt.xlabel('Purchase status')
plt.ylabel('Number of Customers')
plt.title('Total Abandont vs Completed Abandont')
plt.show()

print("----------------Internet Diagnosis----------------")
plt.pie(df['Internet_Speed'],labels=Customers,autopct='%1.1f%%')
plt.title('Internet Diagnosis')
plt.show()

print("----------------Payment Failure Diagnosis----------------")
plt.pie(df['Payment_Failure'],labels=df['Customers'],autopct='%1.1f%%')
plt.title('Payment Failure Diagnosis')
plt.show()

print("----------------Delivery Charges Diagnosis----------------")
plt.hist(df['Delivery_Charges'],bins=8,edgecolor='black')
plt.xlabel('Delivery Charges')
plt.ylabel('Number of Customers')
plt.title('Delivery Charges Diagnosis')
plt.show()

print("----------------Root Cause-------------------")
plt.scatter(df['Internet_Speed'],df['Payment_Failure'])
plt.xlabel('Internet Speed')
plt.ylabel('Payment Failure')
plt.title('Root Cause')
plt.show()
```
