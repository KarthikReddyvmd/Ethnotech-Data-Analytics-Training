```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
import statistics

Days=['Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday']
Water_Usage=[20,25,23,30,28,24,29]
total_water_tank_capacity=35
df=pd.DataFrame({'Days':Days,'Water_Usage':Water_Usage})
print(df)
```

```python
print("Average Of Water Consumption:")
print(df['Water_Usage'].mean())

plt.figure(figsize=(10,5))
plt.bar(df['Days'],df['Water_Usage'])
plt.xlabel('Days')
plt.ylabel('Water Usage')
plt.title('Water Consumption')
plt.show()
```

```python
print("Total Water Consumption in a day:")
for i in range (len(df)):
  day=df['Days'][i]
  usage=df['Water_Usage'][i]
  print(f"{day}: {usage} units")

plt.stem(df['Days'],df['Water_Usage'])
plt.xlabel('Days')
plt.ylabel('Water Usage')
plt.title('Water Consumption')
plt.show()
```

```python
print("Water Usage Change from Monday:")
monday_usage = df['Water_Usage'][0]
for i in range(len(df)):
  day = df['Days'][i]
  usage = df['Water_Usage'][i]
  change = usage - monday_usage
  print(f"{day}: {change} units change")

plt.pie(df['Water_Usage'], labels=df['Days'], autopct='%1.1f%%')
plt.title('Water Consumption Change from Monday')
plt.show()
```

```python
print('Daily Water Consumption Change from Monday:')
monday_usage = df['Water_Usage'][0]
total_change_sum = 0
daily_changes = []
days_for_plot = []
for i in range (len(df)):
  day=df['Days'][i]
  usage=df['Water_Usage'][i]
  change=usage-monday_usage
  total_change_sum += change
  daily_changes.append(change)
  days_for_plot.append(day)
  print(f"{day}: {change} units")

print('Total Sum of Water Consumption Changes from Monday:', total_change_sum)

plt.figure(figsize=(10, 5))
plt.bar(days_for_plot, daily_changes)
plt.xlabel('Days')
plt.ylabel('Water Usage Change from Monday (units)')
plt.title('Daily Water Consumption Change from Monday')
plt.show()
```

```python
prediction=total_change_sum/len(Days)
print("Average of total change sum:",prediction)
final=prediction+df['Water_Usage'][6]
print("Final Water Consumption:",final)
if final>total_water_tank_capacity:
  print("Water Shortage for 8th day.............")
else:
  print("No Water Shortage for 8th day..............")
```
