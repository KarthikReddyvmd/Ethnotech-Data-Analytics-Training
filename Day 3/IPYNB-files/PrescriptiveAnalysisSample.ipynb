```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
import statistics

Patient_names=['Amar','Bob','Jhon','Alice','Karan','Kiran','Vikas','Daniel']
Age=[20,25,60,34,26,48,35,58]
#Symptoms=['Headache','StomachPain','Injury','Heartattack','old',]
Conditions=['Normal','Serious','Major','Serious','Major','Normal','Serious','Normal']
OxygenLevel=[80,90,85,95,82,88,93,84]
Availablebeds=6

df=pd.DataFrame({'Patient_names':Patient_names,'Age':Age,'Conditions':Conditions,'OxygenLevel':OxygenLevel})
print(df)
```

```python
priority1=[]
priority2=[]
priority3=[]
for i in range (len(Patient_names)):
  if Conditions[i]=='Major':
    priority1.append(Patient_names[i])
  elif Conditions[i]=='Serious':
    priority2.append(Patient_names[i])
  elif Conditions[i]=='Normal':
    priority3.append(Patient_names[i])
print('Pririty-1:')
print(priority1,'\n')

print('Pririty-2:')
print(priority2,'\n')

print('Pririty-3:')
print(priority3,'\n')

```

```python
for i in range (len(Patient_names)):
  if OxygenLevel[i] <= 85:
    print('Oxygen level less than 85:')
    print(Patient_names[i])
    print('Shift To ICU Immediately')
  elif OxygenLevel[i] >= 85 and OxygenLevel[i] <90:
    print('Oxygen level greater than 85 and less than 90:')
    print(Patient_names[i])
    print('Patient should be Under observation')
  else:
    print('Oxygen level in normal condition:')
    print(Patient_names[i])
    print('Patient should be Shifted to General Ward')
```

```python
Priority_1=[]
Priority_2=[]
Priority_3=[]
for i in range (len(Patient_names)):
  if OxygenLevel[i] <= 85:
    print('Oxygen level less than 85:')
    print(Patient_names[i])
    print('Shift To ICU Immediately')
    print('\n')
    Priority_1.append(Patient_names[i])
  elif OxygenLevel[i]>= 85 and OxygenLevel[i] <90:
    print('Oxygen level greater than 85 and less than 90:')
    print(Patient_names[i])
    print('Patient should be Under observation')
    print('\n')
    Priority_2.append(Patient_names[i])
  else:
    print('Oxygen level in normal condition:')
    print(Patient_names[i])
    print('Patient should be Shifted to General Ward')
    print('\n')
    Priority_3.append(Patient_names[i])
print('Priority-1:')
print('Shift To ICU Immediately')
print('\n')
print(Priority_1,'\n')

print('Priority-2:')
print('Patient should be Under observation')
print('\n')
print(Priority_2,'\n')

print('Priority-3:')
print('Patient should be Shifted to General Ward')
print('\n')
print(Priority_3,'\n')

```

## Patient Allocation based on Age and Oxygen Level

```python
priority_combined_1 = []
priority_combined_2 = []
priority_combined_3 = []

for i in range(len(Patient_names)):
    if OxygenLevel[i] <= 85 or (OxygenLevel[i] < 90 and Age[i] >= 60):
        priority_combined_1.append(Patient_names[i])
    elif (OxygenLevel[i] >= 85 and OxygenLevel[i] < 90 and Age[i] < 60) or (OxygenLevel[i] >= 90 and Age[i] >= 50):
        priority_combined_2.append(Patient_names[i])
    else: # OxygenLevel >= 90 and Age < 50
        priority_combined_3.append(Patient_names[i])

print('Priority 1 (High Alert - ICU):')
print(priority_combined_1, '\n')

print('Priority 2 (Moderate Alert - Observation):')
print(priority_combined_2, '\n')

print('Priority 3 (Low Alert - General Ward):')
print(priority_combined_3, '\n')
```

```python
Priority_allocated = []
current_available_beds = Availablebeds # Use the initial value of Availablebeds

# Allocate from priority_combined_1
for patient in priority_combined_1:
    if current_available_beds > 0:
        Priority_allocated.append(patient)
        current_available_beds -= 1
    else:
        break # No more beds

# Allocate from priority_combined_2
for patient in priority_combined_2:
    if current_available_beds > 0:
        Priority_allocated.append(patient)
        current_available_beds -= 1
    else:
        break # No more beds

# Allocate from priority_combined_3
for patient in priority_combined_3:
    if current_available_beds > 0:
        Priority_allocated.append(patient)
        current_available_beds -= 1
    else:
        break # No more beds

print("Patients allocated to beds:")
for p_name in Priority_allocated:
  print(p_name)



print(f"\nBeds remaining: {current_available_beds}")
```
