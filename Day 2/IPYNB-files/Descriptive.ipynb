```python
import numpy as np
import statistics
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

member_names=["Aman","Riya","Karan","Sneha","Rahul"]
ages=[22,25,24,23,26]
workout_hours=[2,2.5,2.5,1,3]
calories_burned=[500,350,600,300,750]
print("----------Fitness Center Dataset---------------")
for i in range(len(member_names)):
  print(member_names[i],ages[i],workout_hours[i],calories_burned[i])

average_workout=sum(workout_hours)/len(workout_hours)
print("Average Calories Burned:",average_workout)

highest_calories=max(calories_burned)
print("Highest Calories Burned:",highest_calories)

lowest_workout=min(workout_hours)
print("Lowest Calories Burned:",lowest_workout)

median_calories=statistics.median(calories_burned)
print("Median Calories Burned:",median_calories)

std_dev=statistics.stdev(calories_burned)
print("STD dev:",std_dev)

index=calories_burned.index(highest_calories)
print("Most Active Member:",member_names[index])

index=workout_hours.index(lowest_workout)
print("Least Active Member:",member_names[index])

plt.bar(member_names,calories_burned)
plt.title("Fitness Analysis")
plt.xlabel("Members")
plt.ylabel("Calories")
plt.show()

plt.pie(workout_hours, labels=member_names)
plt.title("Workout Analysis")
plt.show()

plt.hist(calories_burned)
plt.title("Calories Analysis")
plt.xlabel("Calories")
plt.ylabel("Frequency")
plt.show()

plt.scatter(member_names,calories_burned)
plt.title("Calories Analysis")
plt.xlabel("Members")
plt.ylabel("Calories")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb

df=pd.read_csv("exercise_dataset.csv")
print(df.head())

mean_value=df['Calories Burn'].mean()
print("Mean Calories Burned:",mean_value)

median_value=df['Calories Burn'].median()
print("Median Calories Burned:",median_value)

std_value=df['Calories Burn'].std()
print("Standard Deviation Calories Burned:",std_value)

max_value=df['Calories Burn'].max()
print("Maximum Calories Burned:",max_value)

min_value=df['Calories Burn'].min()
print("Minimum Calories Burned:",min_value)

plt.scatter(df['Gender'],df['Calories Burn'])
plt.title("Calories Burned Analysis")
plt.xlabel("Calories Burned")
plt.ylabel("Gender")
plt.show()
```

```python
gender_counts = df['Gender'].value_counts()
plt.pie(gender_counts, labels=gender_counts.index, autopct='%1.1f%%', startangle=90)
plt.title("Distribution of Gender")
plt.axis('equal') # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
```

```python
plt.hist([df['Actual Weight'], df['Dream Weight']], label=['Actual Weight', 'Dream Weight'])
plt.title("Distribution of Actual vs. Dream Weight")
plt.xlabel("Weight")
plt.ylabel("Frequency")
plt.legend()
plt.show()
```

```python
sb.scatterplot(x=df['Calories Burn'],y=df['Actual Weight'])
plt.title("Calories Burned vs. Actual Weight")
plt.xlabel("Calories Burned")
plt.ylabel("Actual Weight")
plt.show()
```

```python
sb.barplot(x=df['Exercise'],y=df['Calories Burn'])
plt.title("Calories Burned by Exercise")
plt.xlabel("Exercise")
plt.ylabel("Calories Burned")
plt.show()
```

```python
sb.regplot(x=df['Actual Weight'],y=df['BMI'])
plt.show()
```

```python
plt.hist(df['Age'], bins=10)
plt.title("Age Distribution")
plt.show()
```

```python
sb.regplot(x='Heart Rate',
           y='Calories Burn',
           data=df)
plt.show()
```

```python
sb.countplot(x='Exercise Intensity',
             data=df)

plt.show()
```
